```python
# ==========================================
# PRAKTIKUM 3 - OPERATOR PERBANDINGAN DAN LOGIKA
# ==========================================

# Tugas 1 & 2: Buat variabel nilai dan kehadiran, gunakan operator >=
nilai = 82
kehadiran = 90

syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

print("--- Kondisi Awal (Nilai: 82, Kehadiran: 90) ---")
print(f"Apakah nilai >= 75? {syarat_nilai}")
print(f"Apakah kehadiran >= 80? {syarat_kehadiran}")

# Tugas 3: Gunakan operator and untuk menentukan kelulusan (kedua syarat harus terpenuhi)
lulus_and = (nilai >= 75) and (kehadiran >= 80)
print(f"Status kelulusan dengan 'and': {lulus_and}")

print("\n" + "="*50 + "\n")

# Tugas 4: Ubah nilai menjadi 70 dan amati perubahan hasilnya
nilai = 70
print(f"--- Setelah Nilai Diubah menjadi {nilai} ---")

syarat_nilai_baru = nilai >= 75
lulus_and_baru = (nilai >= 75) and (kehadiran >= 80)

print(f"Apakah nilai >= 75 sekarang? {syarat_nilai_baru}")
print(f"Status kelulusan baru dengan 'and': {lulus_and_baru}")

# Tugas 5: Coba gunakan operator 'or'
lulus_or = (nilai >= 75) or (kehadiran >= 80)
print(f"Status kelulusan dengan 'or': {lulus_or}")
```
