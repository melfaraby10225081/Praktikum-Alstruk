```python
# ==========================================
# TANTANGAN TERINTEGRASI - PROGRAM REKAP NILAI
# ==========================================

# Fungsi untuk memproses dan menampilkan rekap nilai mahasiswa
def rekap_nilai(nama, nilai_tugas, nilai_ujian):
    # 3. Hitung rata-rata kedua nilai
    rata_rata = (nilai_tugas + nilai_ujian) / 2

    # 4. Mahasiswa dinyatakan lulus jika rata-rata minimal 75
    status_lulus = rata_rata >= 75

    # 5. Tampilkan nama, kedua nilai, rata-rata, dan status lulus
    print(f"Nama Mahasiswa : {nama}")
    print(f"Nilai Tugas    : {nilai_tugas}")
    print(f"Nilai Ujian    : {nilai_ujian}")
    print(f"Rata-rata      : {rata_rata}")
    print(f"Status Lulus   : {status_lulus}")
    print("-" * 40)

# 6. Uji program dengan minimal dua kombinasi nilai yang berbeda
print("=== UJI COBA PROGRAM REKAP NILAI ===\n")

# Kombinasi 1 (Skenario Lulus)
rekap_nilai("Muhammad Elfaraby", 85, 80)

# Kombinasi 2 (Skenario Tidak Lulus)
rekap_nilai("Ahmad", 60, 70)
```
