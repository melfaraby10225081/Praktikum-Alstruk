```python
# ==========================================
# PRAKTIKUM 4 - OPERATOR PENUGASAN
# ==========================================

# 1. Simpan poin awal dalam variabel poin
poin = 100
print(f"Poin awal: {poin}")

# 2. Tambahkan 25 poin menggunakan +=
poin += 25
print(f"Setelah ditambah 25 poin (+=): {poin}")

# 3. Kurangi 10 poin menggunakan -=
poin -= 10
print(f"Setelah dikurangi 10 poin (-=): {poin}")

# 4. Kalikan poin dengan 2 menggunakan *=
poin *= 2
print(f"Setelah dikali 2 (*=): {poin}")

# 5. Bagi poin dengan 2 menggunakan /=
poin /= 2
print(f"Setelah dibagi 2 (/=): {poin}")
```
