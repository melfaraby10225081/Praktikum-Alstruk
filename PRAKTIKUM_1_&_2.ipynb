```python
# Mendefinisikan variabel
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

# 1. Menampilkan semua nilai variabel
print("--- 1. Nilai Variabel ---")
print("Nama:", nama)
print("Umur:", umur)
print("Berat:", berat)
print("Terdaftar:", terdaftar)

# 2. Menampilkan tipe data setiap variabel
print("\n--- 2. Tipe Data Variabel ---")
print("Tipe data nama:", type(nama))
print("Tipe data umur:", type(umur))
print("Tipe data berat:", type(berat))
print("Tipe data terdaftar:", type(terdaftar))

# 3. Mengubah nilai umur menjadi 20
print("\n--- 3. Mengubah Nilai Umur ---")
umur = 20
print("Nilai umur yang baru:", umur)
print("Tipe data umur yang baru:", type(umur))

# 4. Perbedaan 19 dan "19" yaitu :
# Memakai tanda kutip berarti string (data/teks), dan tanpa tanda kutip berarti int (bilangan bulat)
```

```python
# 1. Membuat variabel harga dan jumlah
harga_satuan = 12500
jumlah_beli = 4

# 2. Menghitung total harga
total_harga = harga_satuan * jumlah_beli

# 3. Menghitung sisa pembagian total harga dengan 2.000 menggunakan operator %
sisa_bagi = total_harga % 2000

# 4. Menghitung harga setelah diskon Rp5.000
harga_diskon = total_harga - 5000

# 5. Menampilkan hasil setiap perhitungan dengan label yang jelas
print("Hasil Perhitungan Praktikum 2")
print(f"Harga satuan buku: Rp {harga_satuan:,}")
print(f"Jumlah buku yang dibeli: {jumlah_beli} buah")
print(f"Total harga: Rp {total_harga:,}")
print(f"Sisa pembagian total harga dengan 2.000: Rp {sisa_bagi:,}")
print(f"Harga setelah diskon Rp5.000: Rp {harga_diskon:,}")
```
